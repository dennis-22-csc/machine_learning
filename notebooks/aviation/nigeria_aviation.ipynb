````python
# Check unique values to identify bad entries
print(df1['Q3 2019_ARR_QonQ'].unique())
````

**Most of the values** listed are valid floats (including negative numbers), **but one entry** in the list **is a string**:

```
'-'
```

That is not a negative number. It is just a **single hyphen character**, which pandas reads as a **string**, not a numeric value.

Pandas infers a column's data type based on all entries. If even **one** value is not numeric (like `'-'`), the entire column becomes `object`.

In your unique values list:

```python
[... -82.1576763485477, '-', -4.56690904452098, ...]
                  ↑ this is the culprit
```

---

### ✅ Solution: Clean and Convert

Use `pd.to_numeric()` with `errors='coerce'` to convert the column safely:

```python
df1['Q3 2019_ARR_QonQ'] = pd.to_numeric(df1['Q3 2019_ARR_QonQ'], errors='coerce')
```

This will:

* Convert valid numbers (including negatives) correctly.
* Convert invalid entries like `'-'` to `NaN`.

Then check the new data type:

```python
print(df1['Q3 2019_ARR_QonQ'].dtype)  # Should now be float64
```

````python
df1['Q3 2019_ARR_QonQ'] = pd.to_numeric(df1['Q3 2019_ARR_QonQ'], errors='coerce')
print(df1['Q3 2019_ARR_QonQ'].dtype)  # Should now be float64
````

````python
# Function to test if a column is mostly numeric (e.g., over 50% convertible)
def is_mostly_numeric(series, threshold=0.5):
    converted = pd.to_numeric(series, errors='coerce')
    non_na_ratio = converted.notna().mean()
    return non_na_ratio >= threshold

# Apply only to object columns that are mostly numeric
for col in df1.select_dtypes(include='object').columns:
    if is_mostly_numeric(df1[col]):
        df1[col] = pd.to_numeric(df1[col], errors='coerce')
print(df1.info())  # Get info about the  first sheet
````

````python
file_path = "../../datasets/air_transport_2018_2021_local.csv"
df1 = pd.read_csv(file_path)
````

````python
print(df1.info())
````

````python
# For each quarter and full year, calculate net flow
for period in ['Q1', 'Q2', 'Q3', 'Q4', 'FULL YEAR']:
    for year in [2018, 2019, 2020, 2021]:
        arr_col = f"{period} {year}_ARR"
        dep_col = f"{period} {year}_DEP"
        if arr_col in df1.columns and dep_col in df1.columns:
            df1[f"{period} {year}_NET"] = df1[dep_col] - df1[arr_col]

# Print table of Net Flow for FULL YEAR 2021
net_flow_table = df1[['AIRPORTS', 'FULL YEAR 2021_NET']].copy()
net_flow_table = net_flow_table.sort_values('FULL YEAR 2021_NET', ascending=False)
print("Net Flow by Airport (2021):")
print(net_flow_table.to_string(index=False))

# Visualize net flow by airport
plt.figure(figsize=(12, 6))
sns.barplot(data=net_flow_table, x='AIRPORTS', y='FULL YEAR 2021_NET', palette='coolwarm')
plt.xticks(rotation=90)
plt.title('Net Flow by Airport (2021)')
plt.tight_layout()
plt.show()
````
