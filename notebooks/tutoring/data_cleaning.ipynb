### **1. Introduction (10 minutes)**

#### **What is Data Cleaning?**

Data cleaning (also called data cleansing or scrubbing) is the process of identifying and correcting (or removing) errors and inconsistencies in data to improve its quality and usability.

#### **Why is Data Cleaning Important?**

* Dirty data leads to wrong analysis.
* Machine learning models trained on bad data give poor predictions.
* It improves data accuracy, consistency, and reliability.

#### **2. Common Data Quality Issues (10 minutes)**

| **Issue**                   | **Description**                        | **Example**                         |
| --------------------------- | -------------------------------------- | ----------------------------------- |
| **Missing Values**          | Empty or null values in a dataset.     | Name: John, Age: —                  |
| **Duplicate Records**       | The same entry appears more than once. | Two identical rows.                 |
| **Inconsistent Formatting** | Data is not standardized.              | "Nigeria", "nigeria", "NIGERIA"     |
| **Incorrect Data Types**    | Data stored in the wrong format.       | Age as text: "twenty" instead of 20 |
| **Outliers**                | Unusual or extreme values.             | Income: \$10,000,000                |
| **Typographical Errors**    | Spelling mistakes or wrong entries.    | “Gogle” instead of “Google”         |


#### **3. Techniques for Data Cleaning (20 minutes)**

##### **A. Handling Missing Data**

* Remove rows with missing values.
* Fill missing values with:

  * Mean / Median (for numerical data)
  * Mode / Placeholder (for categorical data)

##### **B. Removing Duplicates**

* Use spreadsheet tools: *Remove duplicates* in Excel/Google Sheets.
* Use Python:

```python
df.drop_duplicates(inplace=True)
```

##### **C. Fixing Inconsistent Formatting**

* Standardize case (upper/lower title).
* Trim white spaces.

Example:

```python
df['country'] = df['country'].str.strip().str.title()
```

##### **D. Converting Data Types**

```python
df['age'] = pd.to_numeric(df['age'], errors='coerce')
```

##### **E. Handling Outliers**

* Use statistical methods (e.g. IQR or Z-score) to detect and possibly remove.
* Visual tools: Boxplot.


#### **4. Practical Example (10 minutes)**

Let’s say we have the following data:

| Name | Age | Country | Income |
| ---- | --- | ------- | ------ |
| John | 25  | Nigeria | 5000   |
| Jane |     | nigeria | 5200   |
| John | 25  | Nigeria | 5000   |
| Mike | 130 | U.S.A   | 10000  |
| Anna | 22  | NIGERIA | —      |

#### Cleaning Steps:

* Fill Jane’s missing age.
* Standardize country names to "Nigeria".
* Remove duplicate row.
* Remove or replace Mike’s age (outlier).
* Handle missing income for Anna.


````python
import pandas as pd
import numpy as np

# Sample data
data = {
    'Name': ['John', 'Jane', 'John', 'Mike', 'Anna'],
    'Age': [25, None, 25, 130, 22],
    'Country': ['Nigeria', 'nigeria', 'Nigeria', 'U.S.A', 'NIGERIA'],
    'Income': [5000, 5200, 5000, 10000, None]
}

df = pd.DataFrame(data)
print("Original Data:")
print(df)



# Cleaning Steps:

# Step 1: Fill Jane’s missing age

# Assume we fill it with the average age (excluding the outlier 130).


# Exclude Mike's age (130) as outlier for mean calculation
filtered_ages = df['Age'][df['Age'] < 100]  # Only normal ages
mean_age = filtered_ages.mean()
df['Age'] = df['Age'].fillna(mean_age)

# Step 2: Standardize country names to “Nigeria”

df['Country'] = df['Country'].str.strip().str.title()

# Unify any known variations
df['Country'] = df['Country'].replace({
    'Nigeria': 'Nigeria',
    'U.S.A': 'USA'
})

# Step 3: Remove duplicate rows

df.drop_duplicates(inplace=True)

# Step 4: Remove or replace Mike’s age (130)

# We'll remove it as it's an unrealistic outlier.

df = df[df['Age'] <= 100]

# Step 5: Handle missing income for Anna

# We’ll fill it with the mean income.

mean_income = df['Income'].mean()
df['Income'] = df['Income'].fillna(mean_income)

# Cleaned Data Output:
print("\nCleaned Data:")
df.reset_index(drop=True, inplace=True)
print(df)

````

````python
import pandas as pd
import numpy as np
import random

# Set random seed for reproducibility
random.seed(42)
np.random.seed(42)

# Sample pool
names = ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve', 'Frank', 'Grace', 'Hannah', 'Ivy', 'Jack']
diagnoses = ['Healthy', 'Diabetes', 'Hypertension', 'Asthma', 'Obesity']

# Generate base clean data
data = {
    'Name': np.random.choice(names, 1900),
    'Age': np.random.randint(18, 90, 1900),
    'BloodPressure': np.random.randint(90, 180, 1900),
    'Diagnosis': np.random.choice(diagnoses, 1900)
}
df = pd.DataFrame(data)

# 1. Introduce Missing Values (10% BloodPressure missing)
num_missing = 200
missing_indices = np.random.choice(df.index, size=num_missing, replace=False)
df.loc[missing_indices, 'BloodPressure'] = np.nan

# 2. Introduce Duplicate Records (5%)
duplicates = df.sample(frac=0.05, random_state=42)
df = pd.concat([df, duplicates], ignore_index=True)

# 3. Introduce Inconsistent Formatting (case variations)
def random_case(text):
    options = [text.lower(), text.upper(), text.capitalize()]
    return random.choice(options)

df['Diagnosis'] = df['Diagnosis'].apply(random_case)
df['Name'] = df['Name'].apply(random_case)

# 4. Incorrect Data Types: Introduce some Ages as text (e.g. 'twenty')
df['Age'] = df['Age'].astype(object)
age_error_indices = np.random.choice(df.index, size=20, replace=False)
for idx in age_error_indices:
    # randomly pick a number word instead of integer
    num_word = random.choice(['twenty', 'thirty', 'forty', 'fifty', 'sixty'])
    df.at[idx, 'Age'] = num_word

# 5. Outliers in BloodPressure (set some extreme values)
outlier_indices = np.random.choice(df.index, size=10, replace=False)
df.loc[outlier_indices, 'BloodPressure'] = np.random.choice([30, 300, 400], size=10)

# 6. Typographical Errors in 'Name' (introduce some misspellings)
def typo(name):
    # Simple typo: duplicate a letter or swap two letters
    if len(name) < 3:
        return name
    pos = random.randint(0, len(name)-2)
    if random.random() < 0.5:
        # duplicate letter
        return name[:pos] + name[pos] + name[pos] + name[pos+1:]
    else:
        # swap letters
        return name[:pos] + name[pos+1] + name[pos] + name[pos+2:]

typo_indices = np.random.choice(df.index, size=30, replace=False)
for idx in typo_indices:
    original_name = df.at[idx, 'Name']
    df.at[idx, 'Name'] = typo(original_name)

# Shuffle rows again
df = df.sample(frac=1, random_state=42).reset_index(drop=True)

# Save to CSV
# df.to_csv("patients.csv", index=False)
````

According to medical guidelines (e.g., WHO, American Heart Association), **normal adult blood pressure** is generally considered to be:

| Category             | Systolic (upper) | Diastolic (lower) |
| -------------------- | ---------------- | ----------------- |
| Normal               | **< 120**        | **< 80**          |
| Elevated             | **120–129**      | **< 80**          |
| Hypertension Stage 1 | **130–139**      | **80–89**         |
| Hypertension Stage 2 | **140+**         | **90+**           |
| Hypertensive Crisis  | **180+**         | **120+**          |

In practice, **a realistic/physiological range** for systolic blood pressure values is roughly:

> **🟢 Normal range: 90–180 mmHg**
> Values **below 80** or **above 200** are often suspect (either outliers or entry errors).

## ✅ What’s Wrong

From the output, we can see:

### 1. **Casing Issues**

* `'Bob'`, `'BOB'`, `'bob'`
* `'Alice'`, `'ALICE'`, `'alice'`, etc.

### 2. **Misspellings / Variants**

* `'frannk'`, `'frrank'`, `'frnak'`, `'FRAKN'`, `'ffrank'` → probably all meant to be `'Frank'`
* `'diiana'`, `'DDiana'`, `'DIANA'` → likely `'Diana'`

* `'charlie'`, `'ccharlie'`, `'CCHARLIE'`, `'Chharlie'` → likely `'Charlie'`

### 3. **Typos / Extra Characters**

* `'AALICE'`, `'aliice'`, `'allice'`, `'AAlice'`, `'ALCIE'` → probably `'Alice'`
* `'Iyv'`, `'IIVY'` → probably `'Ivy'`

## 🔧 How to Clean It


````python
# Step 1: Strip Whitespace and Standardize Case

df['Name'] = df['Name'].str.strip().str.title()

# Step 2: Manually Correct Misspellings

name_corrections = {
    'Frrank': 'Frank',
    'Frnak': 'Frank',
    'Frannk': 'Frank',
    'Ffrrank': 'Frank',
    'Fraank': 'Frank',
    'Frakn': 'Frank',
    'Rfank': 'Frank',
    'Ffrank': 'Frank',

    'Diiana': 'Diana',
    'Ddiana': 'Diana',
    'Diana ': 'Diana',

    'Chharlie': 'Charlie',
    'Ccharlie': 'Charlie',
    'Ccharlie ': 'Charlie',

    'Jakc': 'Jack',
    'Jakc': 'Jack',
    'Ajck': 'Jack',
    'Jacck': 'Jack',

    'Aliice': 'Alice',
    'Allice': 'Alice',
    'Aalice': 'Alice',
    'Alcie': 'Alice',
    'Aaice': 'Alice',

    'Iyv': 'Ivy',
    'Iivy': 'Ivy',
    'Iiivy': 'Ivy',
}

df['Name'] = df['Name'].replace(name_corrections)

# Step 3: Recheck Uniques
print(sorted(df['Name'].unique()))
````

````python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

# Reproducibility
random.seed(42)
np.random.seed(42)

# Data pools
product_clean = ['Widget A', 'Widget B', 'Widget C', 'Widget D']
product_typos = ['widget a', 'Widget A ', 'Widgget B', 'widget C']
region_clean = ['North', 'South', 'East', 'West']
region_typos = ['north ', 'SOUTH', 'eest', 'West ']

# Dates
start_date = datetime.strptime('2023-01-01', '%Y-%m-%d')

# Number of rows
n_rows = 1900  # We'll add duplicates to reach 2000

# Create base data
dates = [start_date + timedelta(days=random.randint(0, 60)) for _ in range(n_rows)]
products = np.random.choice(product_clean + product_typos, size=n_rows)
regions = np.random.choice(region_clean + region_typos, size=n_rows)
sales = np.random.randint(500, 5000, size=n_rows).astype(float)

# Insert some missing values
missing_indices = np.random.choice(n_rows, size=100, replace=False)
for i in missing_indices:
    sales[i] = np.nan

# Insert some outliers
outlier_indices = np.random.choice(n_rows, size=20, replace=False)
for i in outlier_indices:
    sales[i] = np.random.randint(10000, 30000)

# Create DataFrame
df = pd.DataFrame({
    'Date': [d.strftime('%Y-%m-%d') for d in dates],
    'Product': products,
    'Region': regions,
    'Sales': sales
})

# Add duplicates
duplicates = df.sample(frac=0.05, random_state=42)
df = pd.concat([df, duplicates], ignore_index=True)

# Shuffle the data
df = df.sample(frac=1, random_state=42).reset_index(drop=True)

# Save to CSV
df.to_csv("unclean_sales.csv", index=False)
````
