# Numpy Arrays

### Example: Lists vs. NumPy Arrays

````python
import numpy as np

# Python list
list_data = [1, 2, 3, 4]
list_sum = sum(list_data)  # Slow for large data

# NumPy array
np_data = np.array([1, 2, 3, 4])
np_sum = np.sum(np_data)   # Optimized C backend

print(list_sum)
print(np_sum)
````

NumPy arrays can be created from lists or using built-in functions

````python
# 1D Array
arr1d = np.array([1, 2, 3])  

# 2D Array (Matrix)
arr2d = np.array([[1, 2], [3, 4]])  

# Special Arrays
zeros = np.zeros(3)        # [0., 0., 0.]  
ones = np.ones((2, 2))     # [[1., 1.], [1., 1.]]  
range_arr = np.arange(5)   # [0, 1, 2, 3, 4] 
space_arr = np.linspace(0, 1, 5) # [0., 0.25, 0.5, 0.75, 1.]

print(arr1d)
print(arr2d)
print(zeros)
print(ones)
print(range_arr)
print(space_arr)
````

#### 2.2 Indexing and Slicing
##### 1D Arrays (Like Python Lists):

````python
arr = np.array([10, 20, 30, 40])
print(arr[0])      # 10  
print(arr[1:3])     # [20, 30]  
````

##### 2D Arrays (Row, Column Format):

````python
matrix = np.array([[1, 2, 3], [4, 5, 6]])
print(matrix[0, 1])    # 2 (Row 0, Column 1)  
print(matrix[:, 1])    # [2, 5] (All rows, Column 1)  
````

##### Boolean Indexing/Masking:

````python
arr = np.array([1, 2, 3, 4])
print(arr[arr > 2])    # [3, 4]  
````

### 3. Mathematical Operations
#### 3.1 Vectorized Operations
NumPy eliminates the need for loops:

````python
arr = np.array([1, 2, 3])
print(arr * 2)          # [2, 4, 6]  
print(arr + arr)        # [2, 4, 6]  
print(np.sqrt(arr))     # [1., 1.414, 1.732]  
````

#### 3.2 Broadcasting in NumPy

Broadcasting is NumPy's powerful mechanism for performing element-wise operations between arrays of different shapes without explicitly expanding them in memory. It follows strict rules to determine compatibility

---

##### **Key Rules of Broadcasting**  
**Trailing Dimensions Must Match**:  
   - NumPy compares shapes from the **rightmost dimension** backward.  
   - Two dimensions are compatible if:  
     - They are **equal**, or  
     - One of them is **1** (and can be stretched to match).

Broadcasting:

```python
arr + 5  # Adds 5 to each element
arr * 2  # Multiplies each element by 2
``

**Example 1: Vector + Scalar**  

````python
arr = np.array([1, 2, 3])
arr = arr + 5  # Scalar 5 is broadcast to [5, 5, 5]  
# Creates a new array: [1+5, 2+5, 3+5] = [6, 7, 8]

print(arr)
````

**Example 2: Matrix + Vector (Row-wise)**  

````python
matrix = np.ones((3, 3))  # Shape (3, 3)
print(matrix)
vector = np.array([1, 2, 3])  # Shape (3,)
print(vector)

mat_vet = matrix + vector  # Vector broadcast to [[1,2,3], [1,2,3], [1,2,3]]
print(mat_vet)
   
````

**Why is this possible?**  
   - `(3, 3)` + `(3,)` → NumPy aligns trailing dimension (`3` == `3`).  
   - The vector is implicitly treated as a row (shape `(1, 3)`) and **repeated** for all rows.  


**Example 3: Incompatible Shapes**  

````python
matrix = np.ones((3, 3))
print(matrix)
vector = np.array([1, 2])  # Shape (2,)
print(vector)

try:
    mat_vet = matrix + vector  # This will raise a ValueError
    print(mat_vet)
except ValueError as e:
    print(f"Broadcasting failed: {e}")
````

**Error**: `ValueError: operands could not be broadcast together with shapes (3,3) (2,)`  
**Reason**: Trailing dimensions (`3` vs `2`) do not match. 

#### **Explicit Broadcasting with `np.newaxis`**  
If automatic broadcasting fails, reshape arrays manually:  

````python
matrix = np.ones((3, 3))  # Shape (3, 3)
print(matrix)
vector = np.array([1, 2, 3])  # Shape (3,)
print(vector)
vector = vector[:, np.newaxis]  # Reshape to (3, 1)
mat_vet = matrix + vector  # Now broadcasts column-wise
print(mat_vet)
````

##### Tasks:

- Compute mean, median, and standard deviation.

- Normalize the data to a 0–1 scale.

##### Solutions:

````python
data = np.array([22, 45, 34, 19, 28]) 

mean = np.mean(data)  
median = np.median(data)  
std = np.std(data)  
normalized = (data - np.min(data)) / (np.max(data) - np.min(data))  

print(mean)
print(median)
print(std)
print(normalized)
````

Normalizing data to a 0–1 scale means transforming the values so that they fall within the range [0, 1]. This is a common preprocessing step in data analysis and machine learning to ensure that all features are on a similar scale, which can improve the performance of certain algorithms (e.g., gradient descent in neural networks, distance-based methods like k-NN).

##### Theory Behind Min-Max Normalization (0–1 Scaling)
The most common way to normalize data to a 0–1 scale is using **min-max normalization**. The formula is:

$$
x_{\text{normalized}} = \frac{x - \min(X)}{\max(X) - \min(X)}
$$

Where:
- $x$ = original value
- $\min(X)$ = minimum value in the dataset
- $\max(X)$ = maximum value in the dataset
  
##### Steps to Normalize Data
Given data:  
`data = np.array([22, 45, 34, 19, 28])`

1. **Find the min and max**:
   - $ \min(X) = 19 $
   - $ \max(X) = 45 $

2. **Apply the formula**:
   - For $ x = 22 $:  
     $ \frac{22 - 19}{45 - 19} = \frac{3}{26} \approx 0.115 $
   - For $ x = 45 $:  
     $ \frac{45 - 19}{45 - 19} = \frac{26}{26} = 1.0 $
   - For $ x = 34 $:  
     $ \frac{34 - 19}{45 - 19} = \frac{15}{26} \approx 0.577 $
   - For $ x = 19 $:  
     $ \frac{19 - 19}{45 - 19} = \frac{0}{26} = 0.0 $
   - For $ x = 28 $:  
     $ \frac{28 - 19}{45 - 19} = \frac{9}{26} \approx 0.346 $

3. **Result**:  
   The normalized array is approximately:  
   `[0.115, 1.0, 0.577, 0.0, 0.346]`

##### Key Points
- **Purpose**: Ensures all values are on a consistent scale (0–1).
- **When to Use**: Useful when the absolute scale of the data doesn't matter, but relative differences do (e.g., in neural networks, clustering, or distance-based algorithms).
- **Alternative Normalization**: If outliers are present, other methods like **Z-score standardization** (scaling to mean 0, std 1) may be better.
