## 🌟 What Is Pandas?

**Pandas** is an open-source Python library used for data manipulation and analysis. It offers powerful, flexible, and easy-to-use data structures:

* **Series** (1-dimensional)
* **DataFrame** (2-dimensional)

Pandas is built on top of **NumPy** and integrates well with other data science tools.


### 🧱 **1. Series vs. DataFrames**  
**Key Concepts:**  
- **Series:** 1D labeled array (like a column in Excel). A **Series** is a one-dimensional array with labels (axis labels or index). 
  ```python
  import pandas as pd
  s = pd.Series([1, 3, 5, np.nan], index=['a', 'b', 'c', 'd'])
  print(s)

  data = [10, 20, 30]
  series = pd.Series(data, index=['a', 'b', 'c'])
  print(series)

  temps = pd.Series([98.6, 99.1, 97.9], index=['Patient1', 'Patient2', 'Patient3'])
  ```  
- **DataFrame:** 2D table (like an entire Excel sheet). A **DataFrame** is a 2D labeled data structure (like an Excel sheet or SQL table).
 
  ```python
  df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35]
  })
  print(df)
  patients = pd.DataFrame({
      'Age': [45, 32, 67],
      'BP': [120, None, 140],
      'Diagnosis': ['Healthy', 'Diabetes', 'Hypertension']
  })
  ```  

**Comparison:**  
| Feature        | Series                | DataFrame               |  
|---------------|-----------------------|-------------------------|  
| **Dimensions** | 1D                    | 2D                      |  
| **Use Case**   | Single column/vector  | Multi-column datasets   |
