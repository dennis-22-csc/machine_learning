## 🔹 2. Basic Line Plot

### ✅ Example: Plotting a Line

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

plt.plot(x, y)
plt.title("Basic Line Plot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.grid(True)
plt.show()
```

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

plt.plot(x, y)
plt.title("Basic Line Plot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.grid(True)
plt.show()
````

## 🔹 6. Heatmaps

### ✅ Heatmap Example

```python
import matplotlib.pyplot as plt
import numpy as np

# Simulate temperature data for 7 days over 10 weeks
np.random.seed(0)
temperatures = np.random.uniform(low=20, high=35, size=(7, 10))

plt.imshow(temperatures, cmap='hot', interpolation='nearest')
plt.colorbar(label="Temperature (°C)")

# Label axes
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
plt.yticks(ticks=range(7), labels=days)
plt.xticks(ticks=range(10), labels=[f"Week {i+1}" for i in range(10)], rotation=45)

plt.title("Average Daily Temperatures Over 10 Weeks")
plt.xlabel("Week")
plt.ylabel("Day")
plt.tight_layout()
plt.show()

```

````python
import matplotlib.pyplot as plt
import numpy as np

# Simulate temperature data for 7 days over 10 weeks
np.random.seed(0)
temperatures = np.random.uniform(low=20, high=35, size=(7, 10))

plt.imshow(temperatures, cmap='hot', interpolation='nearest')
plt.colorbar(label="Temperature (°C)")

# Label axes
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
plt.yticks(ticks=range(7), labels=days)
plt.xticks(ticks=range(10), labels=[f"Week {i+1}" for i in range(10)], rotation=45)

plt.title("Average Daily Temperatures Over 10 Weeks")
plt.xlabel("Week")
plt.ylabel("Day")
plt.tight_layout()
plt.show()
````

## 🔹 7. Histogram

### ✅ Histogram Example


## 🔹 9. Subplots (Multiple Plots)

### ✅ Using `subplot()`

```python
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
product_A_sales = [150, 200, 250, 220, 300]
product_B_sales = [100, 180, 230, 210, 280]

plt.figure(figsize=(10, 4))

# First subplot: Product A
plt.subplot(1, 2, 1)
plt.plot(months, product_A_sales, marker='o', color='blue')
plt.title("Product A Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)

# Second subplot: Product B
plt.subplot(1, 2, 2)
plt.plot(months, product_B_sales, marker='s', color='green')
plt.title("Product B Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)

plt.tight_layout()
plt.show()
```

````python
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
product_A_sales = [150, 200, 250, 220, 300]
product_B_sales = [100, 180, 230, 210, 280]

plt.figure(figsize=(10, 4))

# First subplot: Product A
plt.subplot(1, 2, 1)
plt.plot(months, product_A_sales, marker='o', color='blue')
plt.title("Product A Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)

# Second subplot: Product B
plt.subplot(1, 2, 2)
plt.plot(months, product_B_sales, marker='s', color='green')
plt.title("Product B Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)

plt.tight_layout()
plt.show()
````

## 🔹 10. Integrating with Pandas

### ✅ Line Plot from DataFrame

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.DataFrame({
    'Year': [2018, 2019, 2020, 2021],
    'Sales A': [200, 220, 250, 300],
    'Sales B': [180, 210, 240, 270]
})

df.plot(x='Year', y=['Sales A', 'Sales B'], kind='line', marker='o')
plt.title("Yearly Sales Comparison")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

````python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.DataFrame({
    'Year': [2018, 2019, 2020, 2021],
    'Sales A': [200, 220, 250, 300],
    'Sales B': [180, 210, 240, 270]
})

df.plot(x='Year', y=['Sales A', 'Sales B'], kind='line', marker='o')
plt.title("Yearly Sales Comparison")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
````
